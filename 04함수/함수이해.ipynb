#### 함수의 정의
- 정의시 def 키워드 사용
- argument도 정의 가능
- 콜론 -> 함수도 코드 블록이므로 꼭 필요함
- body: 함수의 구현 부분

```python
# 함수 정의(define)
# a, b는 매개변수 - 함수 안에서만 참조 가능함.
def add(a, b):
  n = a + b
  return n
```

```python
# 함수 호출
# 함수 호출 시 함수에 전달할 값
res = add(2, 3)
print(res)
```

```python
def add(x, y=10, z=5):
  res = x + y + z
  return res
add(10, 1, 2)
```

return( 리턴 )
- return 옆에 값이나 수식이 있다면 해당 값을 호출자에게반환
- return 만 존재하면 None반환( 프로그램에 대한 안전성 )
- return 이 없는 경우, 기본적으로 함수 코드 블록이 종료 되면 종료로 간주, 이때도 None 반환

복수값 반환
- 튜플을 반환타입으로 사용하면 가능함

```python
def add_mul(x, y):
  s = x + y
  m = x * y
  return s, m
```

```python
a, b = add_mul(2, 3)
print(a, b)
```

가변길이 인자
- 전달되는 파라미터의 갯수가 고정적이지 않은 경우 사용
- print함수, format함수
- *args: 파라미터를 튜플의 형태로 전달
- **kwargs: 파라미털르 딕셔너리 형태로 전달

```python
def test(*args):
  for item in args:
    print(item)
test(10, 20, 30)
```

```python
def test2(**kwargs): # key word arguments
  for key, value in kwargs.items():
    print(key, value)
test2(a=1, name="나신입")
```
