## 람다 함수 이해 및 활용
#### 람다 함수란
- 단일문으로 표현되는 익명함수
- 이름이 없는 구현체만 있는 간단함 함수이다.
- 한 번만 사용되는 기능이 존재할 때 굳이 함수로 만들지 않고 사용함

```python
# 함수도 객체이다. 변수에 저장할 수도 있다.
s = lambda x:x**2 # 익명함수를 만드는 문법
s(5)
```

```python
def s(x):
  return x**2
```

```python
def add(x, y):
  return x + y
add2 = lambda x, y:x+y
add2(10, 20)
```

문자열 리스트를 길이를 기준으로 정렬하기

```python
words = ['hi', 'Nice', 'good time', 'abc']
words.sort(key=lambda s:len(s))
print(words)
```
