```python
import pandas as pd
import folium
import time
from geopy.geocoders import Nominatim
```

```python
df = pd.read_csv('../../data/trade_apt_api_2023_address.txt', sep="\t")
df
```

```python
df['주소'] = '서울시 '+df['지역명']+' '+df['법정동']+' '+df['지번']
print(df['주소'])
```

```python
df
```

```python
geo_local = Nominatim(user_agent='new-apt-map-trainer')
def geocoding(address):
  try:
    geo = geo_local.geocode(address)
    if geo is None:
      return None, None
    return geo.latitude, geo.longitude
  except:
    return None, None
```

```python
N = 30
#N = 3
sample = df.head( N ).copy()
print(sample)
```

```python
sample
```

```python
for i, row in sample.iterrows():
  address = row['주소']
  time.sleep(1)
  lat, lng = geocoding(address)
  print(f'위도: {lat}, 경도: {lng}')
```
