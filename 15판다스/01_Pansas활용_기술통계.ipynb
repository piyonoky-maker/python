## Pandas를 활용한 기술 통계 분석
기술 통계(descriptive statistics)
- 데이터 분포의 특징을 대표할 수 있는 몇 가지 숫자를 계산하여 이 숫자들로 부터 <br>
데이터 분포를 추측하는 방법.

데이터 분포의 특징을 대표하는 값
- 데이터 숫자(count)
- 평균(mean, average)
- 분산(variance)
- 표준편차(standard deviation)
- 최대값
- 최소값
- 중앙값(median)
- 사분위수(quartile)

#### step1 Pandas를 이용한 기술통계 함수
sum(), min(), max(), avg(), describe(), info()함수 제공<br>

```python
import numpy as np
import pandas as pd
```

```python
x = np.array([ 18,   5,  10,  23,  19,  -8,  10,   0,   0,   5,   2,  15,   8,
                2,   5,   4,  15,  -1,   4,  -7, -24,   7,   9,  -6,  23, -13,
                1,   0,  16,  15,   2,   4,  -7, -18,  -2,   2,  13,  13,  -2,
               -2,  -9, -13, -16,  20,  -4,  -3, -11,   8, -15,  -1,  -7,   4,
               -4, -10,   0,   5,   1,   4,  -5,  -2,  -5,  -2,  -7, -16,   2,
               -3, -15,   5,  -8,   1,   8,   2,  12, -11,   5,  -5,  -7,  -4])
```

```python
# Serise 가 여러개 모이면 DataFrame 가 된다.
# describe(): 기술통계에 대한 정보를 한꺼번에 구해주는 함수
se = pd.Series(x)
se
```

count: 결측치를 제외한 갯수
mean: 평균
std: 표준편차
min: 최소값
25%: 1사 분위수(Q1)
50%: 2사 분위수 - 중앙값
75%: 3사 분위수(Q3)
max: 최대값

```python
# 숫자형 Series(se)에 대해 한번에 요약 통계를 Series로 반환
se.describe()
```
