### DataFrame과 리스트, 딕셔너리, 넘파이 dnarray간의 상호 변환<br>
```txt
1. 리스트 -> 데이터 프레임
-> df = pd.DataFrame(list, columns=col_name)
<- DataFrame의 values속성을 이용하여 ndarray 변환: tolist()
2. 딕셔너리 -> 데이터 프레임
-> dict = {'col1': [1, 11], 'col2': [2, 22], 'col3': [3, 33],}
df = pd.DataFrame(dict)
<- 데이터프레임이 저공하는 to_dict()
3. Numpy ndarray -> 데이터 프레임
-> df = pd.DataFrame(array, columns=col_name)
<- 데이터프레임 객체의 values속성을 이용하여 ndarray변환
```

### 데이터 셀렉션 및 필터링
```txt
1. [] : 컬럼 기반 필터링 또는 불린 인덱싱 필터링 제공
  (예:) df.컬럼명, df['컬럼명'], df[1:3], df[:2]
2. loc[인덱스명, 컬럼명] - 명칭 기반 인덱싱
  Index명, column명 기준으로 데이터가 조회
    예:) 하나의 행 검색 : df.loc[0]
        여러 행 검색 : df.loc[[0,1]]
        여러 행과 열 검색 : df.loc[0:2, ['score','grade']]
3. iloc[인덱스, 컬럼인덱스] - 위치기반 인덱싱
  0을 출발점으로 하는 가로축, 세로축 좌표 기반의 행과 열 위치를 기반으로 데이터를 지정
  행,열 위치 값으로 정수를 사용할 수 있음
  주의: 오직 인덱스를 기준으로 데이터를 가져옴. 인덱스가 순서를 의미하지 않음
    예:) df.iloc[0:3], df.iloc[3], df.iloc[0:3, 0:2]
4. 불린(Boolean)인덱싱
  위치 기반, 명칭 기반 인덱싱 모두 사용할 필요없이 조건식을 []안에 기입하여 필터링함
    예:) df.loc[df['score']>90], df.loc[df['grade'] == 'B']
```

````python
# []
data = {
  'name': ['Lee', 'Park', 'Kim'],
  'score': [100, 80, 90],
  'grade': ['A', 'B', 'A']
}
# import pandas as pd
df = pd.DataFrame(data, columns=['name', 'grade', 'score'])
df
````

````python
# 하나의 컬럼 검색
df.name
````

````python
# 2개 이상의 컬럼 검색
df[['name', 'grade']]
````

명칭 기반 인덱싱 -loc[ 인덱스명, 컬럼명 ]

````python
# 하나의 행 검색
df.loc[0]
````

````python
df.loc[[0, 2]]
````

````python
# 여러 행 검색
df.loc[0 : 2]
````

````python
# 여러 행과 열 검색
df.loc[0:2, ['score', 'grade']]
````

위치 기반 인덱싱 - iloc[인덱스, 컬럼인덱스]

````python
df.iloc[0: 3]
````

````python
# 인덱스 임계 숫자는 포함, 컬럼 인덱스 임계는 미포함
df.iloc[0:3, 0:2]
````

````python
# 유효하지 않음 인덱스값은 에러를 발생시킴
df.iloc[3]
````

불린 인덱싱

````python
df.loc[df['score'] > 90]
````
