### Pandas
#### 판다스란?
```txt
패널 데이터 와 Python데이터 분석의 이름을 따서 명명
고수주 자료구조 제공.
Numpy기반에서 개발되어 Numpy사용하는 애플리케이션에 쉽게 적용
```
#### 제공 기능
```txt
통합된 시계열 기능
시계열 데이터와 비시계열 데이터를 함께 다룰 수 있는 통합 자료구조
누락된 데이터를 유연하게 처리할 수 있다.
SQL같은 데이터베이스 처럼 데이터 합치고 관계 연산을 수행하는 기능
```

#### Pandas의 주요 구성 요소 : Series, DataFrame, Index
```txt
Series - 1개 컬럼 값만으로 1차원 데이터셋(Set:집합)
DataFrame - column x row 2차원 데이터 셋
Index - Series와 DataFrame의 고유한 Key값
```

#### Pandas의 주요 함수
```txt
데이터 로드 및 저장 : read_csv(), read_html(), to_csv() 
DataFrame 데이터 확인 : df.shape, df.info(), df.columns, df.head(), df.tail()
각 컬럼마다 Null개수 확인 : df.isnull().sum() 
특정 column / row 삭제 : 
- df = df.drop(['column'], axis=1)
- df = df.drop(['row'])
```
