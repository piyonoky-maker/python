```txt
Missing Data처리
NaN(Not a Number): null과 유사한 개념
1) isna() 또는 isnull(): 주어진 컬럼 값들이 NaN인지 True/False 값을 반환
2) fillna():NaN데이터 찾아서 value지정된 값으로 변환
- df.fillna(value='값', inplace=True)
3) dropna(): 행 데이터를 기준으로 NaN값이 포함된 데이터를 제거
- df.dropna(how='all'): 행의 모든 값이 NaN인 행 데이터 제거
- df.dropna(how='any'): 행 데이터 중 하나라도 NaN이면 해당 행만 제거
- 원본 데이터는 변경되지 않으므로 변경하고 싶다면 inplace=True옵션 추가
```

````python
from sklearn.impute import SimpleImputer
# strategy="mean": 결측치(NaN)를 해당 열의 평균으로 채움(mean, mos_frequent 등도 가능)
imputer = SimpleImputer(strategy="mean")              # 평균으로 대체하기
#imputer = SimpleImputer(strategy="median")           # 중앙값으로 대채하기. 평균과 중앙값이 반드시 일치하는 건 아님
#imputer = SimpleImputer(strategy="most_frequent")    # 최빈값으로 대체하기
#imputer = SimpleImputer(strategy="constant", fill_value=100)    # 특정값으로 대체하기
X[:, 1:3] = imputer.fit_transform(X[: , 1:3])

print(X)
````

````python
# X[:, 1:3] : 2차원 배열(행 X열 ) 슬라이싱
# [:, 1:3]
# 맨 앞 콜론( : )의 경우  행은 전부 선택됨( 즉 모든 행이라는 뜻임 )
# 두번째는 열 인덱스 1 이상 3미만 -> 2번째 열, 3번째 열 ( 총 2개 열 출력함)
print(X[:, 1:3])
````
