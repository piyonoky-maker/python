## 평균 제곱 오차
```txt
최소 제곱법을 이용해서 기울기와 y절편을 구했지만 이 공식만으로
앞으로 만나게 될 모든 상황을 해결하기는 어렵다.
왜냐면 여러 개의 입력을 처리해야 하는 예를 들어 폐암 수술 환자의 생존율 데이터를 보면
입력 데이터의 종류가 17개나 된다. 
딥러닝은 대부분 입력 값이 여러 개인 상황에서 이를 해결하기 위해 실행되기 때문에
다른 방법이 필요함.

가설을 하나 세운 후 이 값이 주어진 요건을 충족하는지 판단해서 조금씩 변화를 주고
이 변화가 긍정적이면 오차가 최소가 될 때 까지 이 과정을 계속하는 반복하는 방법이
딥러닝을 가능하게 하는 가장 중요한 원리 중 하나 임. 
```

````python
import numpy as np
````

````python
# 가상의 기울기 w와 y절편 b를 정함
fake_w = 3
fake_b = 76
````

````python
# 데이터 값을 세팅함
# 공부시간 x와 성적 y의 넘파이 배열 생성
x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97])
````

````python
# y = wx + b
def predict(x):
  return fake_w * x + fake_b
````

````python
# 회귀식으로 각 x에 대한 예측값 구하기
# 예측값을 담을 리스트 선언( 평균제곱오차(MSE)와 쌍으로 사용 )
predict_result = []
# 공부시간 마다 회귀식 ( predict )을 적용한 예측점수 구함
for i in range(len(x)):
  predict_result.append(predict(x[i]))
  print('공부시간=%.f, 실제점수=%.f, 예측점수=%.f' %(x[i], y[i], predict(x[i])))
````

````python
# MSE(평균제곱오차) 함수를 각 y값에 대입해서 최종값을 구하는 함수 선언
n = len(x)
def mse(y, y_pred):
  return (1/n) * sum((y - predict_result)**2)
# 평균 제곱 오차 값을 출력하기
print(f'평균 제곱 오차 : {str(mse(y, predict_result))}')
````
