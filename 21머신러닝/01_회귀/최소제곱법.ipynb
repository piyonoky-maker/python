## 최소제곱법

```python
import numpy as np
```

공부한 시간과 점수를 각각 x, y라는 이름의 넘파이 배열로 만든다

```python
# 데이터 값을 세팅함
x = np.array([1, 2, 3])
y = np.array([3, 5, 7])
```

```python
mx = np.mean(x)
my = np.mean(y)
```

```python
print(mx)
print(my)
```

기울기 구하기<br>
기울기 공식에서 분모 부분과 분자 부분을 구한다<br>

```python
# 기울기 공식의 문모 부분
divisor = sum([(i-mx)**2 for i in x])
divisor
```

```python
# 기울기 공식의 분자 부분
def top(x, mx, y, my):
  d = 0
  for i in range(len(x)):
    d += (x[i] - mx) * (y[i] - my)
  return d
dividend = top(x, mx, y, my)
dividend
```

```python
# 기울기 w를 구하는 공식
w = dividend / divisor
```

```python
# y절편 b를 구하는 공식
b = my - (mx*w)
b
```
