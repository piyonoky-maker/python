html문자열 파싱

````python
html = '''
<html>
  <head>
    <title>BeautifulSoup test</title>
  </head>
  <body>
    <div id='upper' class='test' custom='good'>
      <h3 title='Good Content Title'>Contents Title</h3>
      <p>Test contents</p>
    </div>
    <div id='lower' class='test' custom='nice'>
      <p>Test Test Test 1</p>
      <p>Test Test Test 2</p>
      <p>Test Test Test 3</p>
    </div>
  </body>
</html>'''
````

find함수<br>
- 특정 html tag를 검색
- 검색 조건을 명시하여 찾고자 하는 tag

````python
from bs4 import BeautifulSoup
soup = BeautifulSoup(html)
````

````python
soup.find("h3")
````

````python
soup.find("p")
````

````python
soup.find("div")
````

````python
soup.find("div", class_="test")
````

````python
attrs = {'id': 'upper',  'class': 'test'}
soup.find('div', attrs = attrs)
````

find_all함수
- 조건에 맞는 모든 tag를 리스트로 반환함

````python
soup.find_all('div', class_='test')
````

get_text함수
- tag안의 value를 추출
- 부모 tag의 경우 모든 자식 tag의 value를 추출

````python
tag = soup.find('h3')
print(tag)
tag.get_text()
````

````python
tag = soup.find('p')
print(tag)
tag.get_text()
````

attribute 값 추출하기
- 경우에 따라 추출하고자 하는 값이 attribute 에도 존재
- 검색한 tag에 attribute이름을 []연산으로 추출가능함
- 예) div.find('h3')['title']

#### 다음 뉴스 데이터 추출하기
뉴스기사에서 제목, 작성자, 작성일, 좋아요 등등
- id, class속성으로 tag찾기

````python
import requests
url = 'https://v.daum.net/v/20260213095214358'
res = requests.get(url)
res.text
````

````python
soup = BeautifulSoup(res.text)
title = soup.find('h3', class_='tit_view')
title
````

````python
title.get_text()
````

````python
writer = soup.find_all('span', class_='txt_info')[0]
writer
````

````python
second = soup.find_all('span', class_='txt_info')[1]
second
````

````python
data_value = soup.find('span', class_='num_date')
print(data_value)
data_value.get_text()
````

크롬열기 -> 유튜브접속 -> 요소를 찾는다. -> 5초간 기다림 -> 후처리

````python
# 브라우저를 자동으로 조작
from selenium import webdriver
# 요소를 찾는 방법(ID, XPATH, CSS 등)
from selenium.webdriver.common.by import By
# 잠깐 동안 기다리기(로딩용)
import time
# 브라우저 실행
driver = webdriver.Chrome()
driver.get('https://www.youtube.com/')
searchs = driver.find_elements(By.XPATH,'//*[@id="center"]/yt-searchbox/div[1]/div/form/input')
print(searchs)
time.sleep(5)
````
