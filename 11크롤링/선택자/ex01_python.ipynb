#### 네이버 검색창에 인공지는 입력 후 조회 결과 페이지 출력하기
send_keys(): 셀레니옴이 제공하는 함수임

````python
# 브라우저 실행
driver = webdriver.Chrome()
# 접속
driver.get('https://naver.com')
# 네이버 검색창 선택한 후 오른쪽 버튼 -> 복사 -> xpath복사
search = driver.find_element(By.XPATH, '//*[@id="query"]')  # 검색창에 대한 부분 접근

# 입력값 초기화
search.clear()
search.send_keys("인공지능")  # send_keys는 셀레니엄이 제공하는 함수

# 입력키에 단어 넣고 검색 버튼 누르기
# query.send_keys(Keys.RETURN)

# 동작
# 돋보기 모양 버튼 접근
search_button = driver.find_element(By.XPATH, '//*[@id="sform"]/fieldset/button')
search_button.click()

time.sleep(5)
driver.close()
````

만일 Enter키 입려긍로 검색어를 입력하고 싶다면<br>

함수 파라미터에 Keys.RETURN을 하면 키를 누른 것과 같은 효과를 연출할 수 있음

````python
from selenium.webdriver.common.keys import Keys

# 브라우저 실행
driver = webdriver.Chrome()
# 접속
driver.get('https://naver.com')
# 네이버 검색창 선택한 후 오른쪽 버튼 -> 복사 -> xpath복사
search = driver.find_element(By.XPATH, '//*[@id="query"]')  # 검색창에 대한 부분 접근

# 입력값 초기화
search.clear()
search.send_keys("인공지능")  # send_keys는 셀레니엄이 제공하는 함수

# 입력키에 단어 넣고 검색 버튼 누르기
# query.send_keys(Keys.RETURN)

# 동작
# 돋보기 모양 버튼 접근
search.send_keys(Keys.RETURN)

time.sleep(5)
driver.close()
````

### Youtube 검색 결과 수집하기
- https://youtube.com/
1. 유튜브 접속
2. 검색창에 '인공지능' 입력
3. Enter로 검색
4. video-title 요소 텍스트 출력하기

````python
# 첫번째 실습
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

driver = webdriver.Chrome()
driver.get("https://www.youtube.com/")
search_input = driver.find_element(By.XPATH, '//*[@id="center"]/yt-searchbox/div[1]/div/form/input')
search_input.send_keys("인공지능")
time.sleep(3)
# 엔터키 입력처리
search_input.send_keys(Keys.RETURN)
time.sleep(10)
titles = driver.find_elements(By.XPATH ,'//*[@id="video-title"]' )
for title in titles:
  print(title.text)

driver.quit()
````
