#### 식사비용 계산하기
팁금액 = 식사비용*인원 / 팁비용

1. 사용자로부터 총 식사 금액과 팁 비율를 입력 받는다
2. 팁 금액을 계산 합니다
3. 최종 금액을 계산
4. 인원 수를 정의
5. 각 사람이 내야 할 금액을 계산
6. 결과를 출력

```python
total_bill = float(input("총 식사 금액을 입력하세요. 예: 100000"))
tip_percentage = float(input("원하는 팁의 비율울 입력하세요 (예: 10, 15, 20)"))
```

```python
print(total_bill, tip_percentage)
```

최종 금액을 계산합니다

```python
tip_amount = (total_bill * tip_percentage) / 100
print(tip_amount)
```

```python
final_bill = total_bill + tip_amount
```

인원 수를 정의 합니다 ( 이번 경우에는 5명 으로 )

```python
number_of_people = 5
```

각 사람이 내야 할 금액을 계산합니다.

```python
amount_per_person = final_bill / number_of_people
print(amount_per_person)
```

결과를 출력합니다

```python
print(f"총 팁 금액 : {tip_amount} 원")
print(f"총 최종 금액 : {final_bill} 원")
print(f"각 사람이 내야 할 금액 : {amount_per_person} 원")
```
