### 기본 자료형
#### 자료형( 데이터 타입 )
- 프로그래밍 언어들은 변수가 가져야 할 값의 크기, 형식, 범위 등에 따라서 값의 유형을 정함.<br>
- 자료형은 프로그래밍 언어에서 정수, 실수, 논리 등  여러 종류의 자료가 어떤 값을<br>
가질 수 있는 지에 대해 알려주는 속성.<br>
- 자료형에 따라 저장할 수 있는 값의 범위( 크기 )와 값을 저장하는 방식이 달라짐.<br>
- 자료형에 따라 수행할 수 있는 명령들이 달라 질 수 있음<br>

자바나 C언어와 차이점은<br>
- 파이썬은 자료형의 개념이 있지만 변수를 선언할 대 자료형을 지정하지 않음.<br>

숫자형, 문자형, 이스케이프 문자, 유니코드, 논리형, 문자열, 자료형과 관련 함수들..<br>

#### 포맷팅
- 포맷팅: 문자, 숫자, 날짜 데이터에  형식을 지정하는 것<br>
printf<br>
format(): '{[인데스]: [공백대체문자][정렬방법][자릿수][타입]}'.format(변수)
- 인덱스 : format() 함수의 인수 중에서 해당 자이에 출력할 인수의 인덱스
- 정렬방법 : <기호는 왼쪽 정렬, > 기호는 오른쪽 정렬, ^기호는 가운데 정렬 해서 출력
- 자릿수 : 변수의 값을 출략할 최대 최대 자릿수를 지정
- 타입 : 출력 형식을 지정( d는 10진정수, f는 실수, s는 문자열, b는 2진수, o는 8진수, x는 16진수 )

````python
name = '김유신'
age = 20
print(name, '님의 나이는 ', age, '세 입니다', sep='')
````

````python
print('{}님의 나이는 {}세 입니다.'.format('김유신', 20))
````

````python
print('{}님의 나이는 {}세 입니다.'.format(name, age))
````

````python
# {}에 인덱스를 지정해서 매핑 시킴
print('출력: {2}, {0}, {1}'.format(10, 20, 30))
````

````python
a = 12345
print('출력 : [{}],[{:10}],[{:3}]'.format(a, a, a))
````

````python
a = 12345
print('출력 : [{}],[{:10}],[{:3}]'.format(a, a, a))
````

````python
a = 12345
print('출력 : [{:d}],[{:f}],[{:b}],[{:o}],[{:x}]'.format(a, a, a, a, a))
````

````python
a = 1234567.896
print('출력 : [{}],[{:f}],[{:10f}],[{:10.2f}],[{:20.10f}]'.format(a, a, a, a, a))
````

- 문자 자료형 형식 지정<br>
{:width.presion} : width: 전체 자리수, precision: 출력할 문자의 개수<br>

````python
b = 'Hello World'
print('출력 : [{}], [{:s}], [{:20}]'.format(b, b, b))
````

#### 정렬 방법 지정
{}의 콜론 뒤에 '<', '>', '^' 을 이용하면 정렬 상태를 지정.

````python
a, b, c = 10, 20, 30
print('출력 : [{:>6d}], [{:<6d}],[{:^6d}]'.format(a, b, c))
````

#### f-strings
- 3.6버전 부터 지원
- 문자열에 형식을 지정할 때 서식을 쉽게 만들 수 있도록 함.

````python
data = "2026-02-11"
year, month, day = data.split('-')
print(f"연도: {year}, 월: {month}, 일: {day}")
````
