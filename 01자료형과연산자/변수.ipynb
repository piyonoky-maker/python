#### 변수
프로그램이 실행되는 동안 상황에 따라 변하는 값을 저장.<br>

변수 사용 이점
- 이름을 부여하고 값을 저장해 놓으면 연산에 변수 이름을 사용할 수 있다.<br>
변수 선언<br>
- 값을 저장할 수 있는 변수를 만드는 것<br>
- 파이썬은 변수 선언하기 위한 타입이 없으므로 변수 선언 과정이 없음.<br><br>
변수 할당<br>
-  어떤 값이 이름을 갖도록 하는 과정<br>
- =연산자를 이용함<br>
- 변수이름 = 연산_결과_값<br>
변수는 파이썬에서 가장 많이 사용되는 객체임

변수 이름 규칙<br>
- 변수의 이름 문자, 숫자, 밑줄(_)문자를 포함하며, 대/소문자를 구분함.<br>
- 숫자로 시작할 수 없으며, 밑줄 제외한 특수문자, 공백, 문장부호 사용불가<br>
- 파이썬의 예약어는 사용불가<br>

- 세계 대부분의 언어를  문자열 리터럴, 식별자 및 주석에 사용할 수 있음
- 한글을 처리하기 위해 utf-8을 사용하므로 별도로 인코딩을 지정하지 않아도 됨

인코딩 이란?
```text
인코딩은 정보의 형태나 형식을 변화하는 처리나 처리방 식을 의미함.
여기서 사용한 인코딩은 문자 인코딩을 의미.
문자 인코딩을 잘못 지정했을 경우 한글이 깨짐 현상 발생.
예) UTF-8, EUC-KR, MS949
```

#### 사용자 입력
input(): 사용자로 부터 값을 입력 받기 위한 함수, 입력 받은 값은 항상 문자열임.
문자열을 숫자로 변환하는 함수: int()함수

````python
# 두 수를 입력 받아서 두 수의 합을 출력하시오
first = int(input('첫번째 숫자를 입력하시오: '))
print(first)   # 2 -> "2"
print(type(first))   #
````

````python
second = int(input('두번째 숫자를 입력하세요'))
print(first+second)
````

예제 #1: 변수를 이용한 조건문

````python
# temperature = 18  #온도를 나타내는 변수
temperature = int(input('온도를 입력해주세요'))
if temperature > 25:
  print('hot')
elif temperature >= 15:
  print('nice')
else:
  print('cold')
````

예제 #2: 문자열/ 리스트/ 릭셔너리 타입 체크

````python
name = '노트북'
items = [1, 2, 3]
product = {'name': '키보드', 'price': 10000}
print(type(name))
print(type(items))
print(type(product))
````

````python
print(isinstance(name, str))
print(isinstance(items, list))
print(isinstance(product, dict))
````
