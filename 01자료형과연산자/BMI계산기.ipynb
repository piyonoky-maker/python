### BMI 계산기 구현
1. 학습목표
- input() 으로 값을 입력 받고 문자열을 숫자형으로 변환할 수 있다
- 산술 연산을 사용해 BMI공식을 작성할 수 있다..
- 계산 결과를 print()로 출력할 수 있다
- 정수/실수  타입 차이를 이해하고 적용할 수 있다

2. 문제 설명
- BMI공식을 검색하여 공식을 세운다
- 키는 미터 단위로 입력한다.
- 몸무게는 Kg단위로 입력한다.

3. BM지수 출력해 보기

4. 판정하기

```python
height = float(input("당신의 키를 알려주세요 m: "))
weight = float(input("당신의 무게를 알려주세요 kg:"))
```

```python
BMI = int(weight)/float(height)**2
```

```python
bmi_as_int = int(round(BMI))
print(bmi_as_int)
```

```python
if bmi_as_int < 18.5:
  print(f"당신의 BMI는 {bmi_as_int}, 당신은 저체중 입니다")
elif bmi_as_int < 25:
  print(f"당신의 BMI {bmi_as_int}, 당신은 정상 입니다")
elif bmi_as_int < 30:
  print(f"당신의 BMI {bmi_as_int}, 당신은 경도 비만 입니다")
elif bmi_as_int < 35:
  print(f"당신의 BMI {bmi_as_int}, 당신은 비만 입니다")
else:
  print(f"당신의 BMI {bmi_as_int}, 당신은 과체중 입니다")
```
