```python
import requests

# API 요청 URL
url = 'https://openapi.naver.com/v1/search/book.json?query=파이썬&start=1&display=3'
# 요청 헤더에 필요한 Client ID와 Client Secret 추가
headers = {
  'X-Naver-Client-Id': 'viTHjLZxdPpT1RKMsi8w',
  'X-Naver-Client-Secret': 'oheRHD0lHz'
}
#HTTP 요청을 보내고 응답을 받아오기
res = requests.get(url, headers = headers)
print(res.json()['items'])
```
