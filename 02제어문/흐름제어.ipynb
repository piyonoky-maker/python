#### 키/나이를 입력받아 탑승 여부와 요금을 계산하는 프로그램

#### 1) 키 입력

```python
height = int(input('키(cm)를 입력하세요. :'))
print(height)
```

#### 2) 판정, 다시 질문, 판정

```python
if height > 120:
  print('탑승 가능합니다.')
  # 3) 나이 검사
  age = int(input('나이를 입력하세요..'))
  if age <= 7:
    print('요금은 $7 입니다.')
  else:
    print('요금은 $12 입니다.')
else:
  print('탑승이 불가합니다')
  
```

## step윤년: 흐름 제어
<img src="./윤년.png" width="500px">

```python
# 윤년인지  확인하고 싶은 숫자를 입력하세요.
year = int(input("윤년인지 확인하고 싶은 숫자를 입력하세요"))
year
```

```python
if year % 4 == 0:
  if year % 100 == 0:
    if year % 400 == 0:
      print(f"{year} is a leap year")
    else:
      print(f"{year} is not a leap year")
  else:
    print(f"{year} is a leap year")
else:
  print(f"{year} is a leap year")
```
