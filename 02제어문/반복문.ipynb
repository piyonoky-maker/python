range(from, to, by) # 사용빈도가 많은 함수
- : range()를 이용하면 인덱스 위주의 반복을 실행 시킬 수 있다

```python
# 세번째 파라미터는 얼마씩 증가시킨 값을 갖게 할 것인지를 결정함
# 생략이되면 1씩 증가
for i in range(0, 10, 1):
  print(i, end='\t')
```

```python
내취미 = ['낚시', '독서', '쇼핑']
for 내취미 in 내취미:
  print('%s 는 내가 좋아하는 취미 입니다' %내취미)
```

```python
# 딕셔너리는 itelms(). keys(), values()함수를 사용함
color = {'apple': '사과', 'banana': '바나나', 'kiwi': '키위'}
for c in color.items():
  print(c)
```

#### 반복문 ( while )
조건문이 참일 동안 계속 실행<br>
- 조건문을 검사해서 결과가 참( True ) 일 동안 블록을 반복 실행함.

```python
i = 0
while i < 10:
  print(i, end = ' ')
  i += 1
```

```python
i = 0
while i < 10:
  if i %2 == 0:
    print(i, end = ' ')
  i += 1
```

```python
i = 0
while i < 10:
  if i %2 == 1:
    print(i, end = ' ')
  i += 1
```

```python
# break를 만나면 break를 포함하는 반복문을 탈출
i = 0
while i < 10:
  if i == 5:
    break
  print(i, end = ' ')  
  i += 1
```

```python
# continue는 반복문에서 continue 이후의 문장을 건너뜀
i = 0
while i < 10:
  i += 1
  if i == 5:
    continue
  print(i, end = ' ') 
```
